Upload data to the Qdrant

```python
from config import Config
cfg = Config()
```

```python
from ingest.embed import build_index

embeds = build_index(cfg)
```

```python
from qdrant_client import QdrantClient
client = QdrantClient(url="http://localhost:6333")
```

```python
from qdrant_client.models import Distance, VectorParams

client.create_collection(
    collection_name="textbook_corpus",
    # MedCPT-Article-Encoder's [CLS] output is 768-dim; DOT because the model's
    # own usage example scores raw, unnormalized vectors (not cosine).
    vectors_config=VectorParams(size=768, distance=Distance.DOT),
)
```

```python
import numpy as np
from tqdm import tqdm
from qdrant_client.models import PointStruct
from pathlib import Path
from ingest.embed import load_corpus

COLLECTION_NAME = "textbook_corpus"
CACHE_DIR = Path(cfg.cache_embed)

#load vector and payload
dataset = load_corpus(cfg)
cache_files = sorted(CACHE_DIR.glob("batch_*.npy"))

row = 0
for cache_file in tqdm(cache_files, desc="upserting", unit="batch"):

    #load file as numpy - 2d array of (64, 768)
    embeds = np.load(cache_file)

    #build batch
    batch_len = embeds.shape[0]
    records = dataset[row : row + batch_len]  # same lo/hi range build_index used to make this batch

    #format points for upsert
    points = [
        PointStruct(
            id=row + i,  # stable row index -> re-running this cell overwrites instead of duplicating
            #1D array of (768,)
            vector=embeds[i].tolist(),
            payload={
                "text": records["content"][i],
                "title": records["title"][i],
                "chunk_id": records["id"][i],
                "source": cfg.corpus,
            },
        )
        for i in range(batch_len)
    ]

    #upsert to qdrant
    client.upsert(collection_name=COLLECTION_NAME, points=points)
    row += batch_len

assert row == len(dataset), f"upserted {row} rows but dataset has {len(dataset)}"
print(f"done: {row} points upserted into '{COLLECTION_NAME}'")
```
